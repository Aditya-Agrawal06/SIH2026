```python
import time
import requests
from pathlib import Path

def download_imd_rain(years, out_dir="../data/raw/rain", max_retries=4, timeout=60):
    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)
    url = "https://imdpune.gov.in/cmpg/Griddata/rainfall.php"

    for year in years:
        fname = out_dir / f"{year}.grd"
        if fname.exists() and fname.stat().st_size >= 1024:
            print(f"{year}: already have it, skipping")
            continue

        for attempt in range(1, max_retries + 1):
            try:
                print(f"{year}: attempt {attempt}/{max_retries}")
                resp = requests.post(url, data={"rain": year}, timeout=timeout)
                resp.raise_for_status()
                if len(resp.content) < 1024:
                    raise ValueError("response too small, likely an error page")
                fname.write_bytes(resp.content)
                print(f"{year}: saved ({len(resp.content)/1e6:.1f} MB)")
                break
            except Exception as e:
                wait = 15 * attempt
                print(f"{year}: failed ({e}), retrying in {wait}s")
                time.sleep(wait)
        else:
            print(f"{year}: FAILED after {max_retries} attempts — skipping for now")

download_imd_rain(range(2015, 2024))
```

```python
import pandas as pd

rain_df = pd.read_csv(
    "../data/processed/periyar_daily_rainfall_2015.csv",
    parse_dates=["time"]
)

rain_df = rain_df.sort_values("time").reset_index(drop=True)

print(rain_df.head())
```

```python
rain_df["rainfall_3day"] = (
    rain_df["rainfall_mm"]
    .rolling(window=3)
    .sum()
)

rain_df["rainfall_7day"] = (
    rain_df["rainfall_mm"]
    .rolling(window=7)
    .sum()
)

rain_df["rainfall_15day"] = (
    rain_df["rainfall_mm"]
    .rolling(window=15)
    .sum()
)

rain_df["rainfall_3day_max"] = (
    rain_df["rainfall_mm"]
    .rolling(window=3)
    .max()
)

rain_df["rainfall_7day_max"] = (
    rain_df["rainfall_mm"]
    .rolling(window=7)
    .max()
)
```

```python
rain_df["rainfall_previous_day"] = (
    rain_df["rainfall_mm"].shift(1)
)
```

```python
print(rain_df.head(20))
```

```python
print(rain_df.isnull().sum())
```

```python
rain_df = rain_df.dropna().reset_index(drop=True)

print("Shape after removing incomplete rows:", rain_df.shape)
print()
print(rain_df.isnull().sum())
```

```python
feature_path = "../data/processed/periyar_rainfall_features_2015.csv"

rain_df.to_csv(feature_path, index=False)

print("Feature dataset saved!")
print(feature_path)
```
